```python
# Datos del enunciado
rf = 0.02          # tipo libre de riesgo
rm = 0.09          # rentabilidad esperada del mercado
sigma_m = 0.10     # desviación típica del mercado

# 1) Pesos en activo sin riesgo y en cartera de mercado para E[R] = 12,5%
E_target = 0.125

w_m = (E_target - rf) / (rm - rf)   # peso en cartera de mercado
w_rf = 1 - w_m                      # peso en activo sin riesgo

print("1) w_rf (activo sin riesgo) =", w_rf)   # → -0.5
print("   w_m  (cartera de mercado) =", w_m)   # → 1.5

# 2) Riesgo (desviación típica) de esta cartera
sigma_p = abs(w_m) * sigma_m       # solo el activo riesgoso aporta varianza
print("2) sigma_p =", sigma_p)     # → 0.15

# 3) Correlación entre la cartera del apartado 2 y el mercado
# Como es combinación lineal de rf + mercado, la correlación es ±1
corr_pm = 1 if w_m > 0 else -1
print("3) correlación cartera-mercado =", corr_pm)   # → 1

# 4) Precio de equilibrio del riesgo
# a) Por unidad de desviación típica (Sharpe del mercado)
precio_riesgo_sigma = (rm - rf) / sigma_m

# b) Por unidad de varianza (λ del CAPM)
precio_riesgo_var = (rm - rf) / (sigma_m ** 2)

print("4) Precio del riesgo (por sigma)   =", precio_riesgo_sigma)  # → 0.7
print("   Precio del riesgo (por sigma^2) =", precio_riesgo_var)    # → 7.0

# 5) Alpha de Jensen si beta de la cartera es 0.67
beta_p = 0.67
E_capm = rf + beta_p * (rm - rf)   # rendimiento que predice CAPM
alpha_jensen = E_target - E_capm   # alpha = rendimiento real - CAPM

print("5) Alpha de Jensen =", alpha_jensen)    # ≈ 0.0581

# 6) (Opcional) Comentario CAPM vs APT usando el alpha calculado
if abs(alpha_jensen) > 0:
    print("6) Alpha distinto de 0 → CAPM no explica totalmente el rendimiento; "
          "podría preferirse un APT con un factor adicional.")

# 7) Beta de un activo con rentabilidad esperada del 15%
R_expected = 0.15
beta_7 = (R_expected - rf) / (rm - rf)

print("7) Beta del activo con E[R]=15% =", beta_7)   # ≈ 1.8571

# 8) Beta de una cartera equiponderada de dos activos (betas 1.8 y 0.6)
beta1, beta2 = 1.8, 0.6
beta_port = 0.5 * beta1 + 0.5 * beta2

print("8) Beta cartera equiponderada =", beta_port)  # → 1.2




```

```python
# EJERCICIO 2
# Coeficientes (lambdas)
lam = {
    "lam0": 0.021,
    "lam1": 0.045,   # mercado español
    "lam2": 0.032,   # mercado europeo
    "lam3": 0.006,   # situación económica
    "lam4": -0.004,  # divisas
    "lam5": 0.348    # pendiente ETTI
}

# Desviaciones típicas de los coeficientes
se = {
    "lam0": 0.25,
    "lam1": 0.018,
    "lam2": 0.95,
    "lam3": 0.003,
    "lam4": 0.0016,
    "lam5": 0.58
}

# p-valores de cada coeficiente
pvals = {
    "lam0": 0.223,
    "lam1": 0.005,
    "lam2": 0.351,
    "lam3": 0.031,
    "lam4": 0.017,
    "lam5": 0.567
}

R2 = 0.37
R2_adj = 0.239
F_stat = 2.82
F_pval = 0.038

# ============================
# 1) ESTADÍSTICOS t PEDIDOS
# ============================

t_lam4 = lam["lam4"] / se["lam4"]    # riesgo divisas
t_lam5 = lam["lam5"] / se["lam5"]    # pendiente ETTI

print("1) t para riesgo en mercado de divisas (λ4):", t_lam4)
print("1) t para riesgo por cambios en la pendiente de la ETTI (λ5):", t_lam5)

# ============================
# 2) FRASES DE RELEVANCIA
# ============================

# a) “El riesgo del mercado español es irrelevante
#     para explicar el rendimiento esperado a
#     cualquier nivel de significación superior al 1%”
alpha_1 = 0.01
frase_a = pvals["lam1"] > alpha_1
print("\n2a) La frase sobre el mercado español es", 
      "VERDADERA" if frase_a else "FALSA")

# b) “El riesgo del mercado europeo es irrelevante
#     para explicar el rendimiento esperado a
#     cualquier nivel de significación superior al 10%”
alpha_10 = 0.10
frase_b = pvals["lam2"] > alpha_10
print("2b) La frase sobre el mercado europeo es", 
      "VERDADERA" if frase_b else "FALSA")

# c) “El riesgo asociado a la situación económica es irrelevante
#     para explicar el rendimiento esperado a cualquier nivel
#     de significación superior al 5%”
alpha_5 = 0.05
frase_c = pvals["lam3"] > alpha_5
print("2c) La frase sobre la situación económica es", 
      "VERDADERA" if frase_c else "FALSA")

# ============================
# 3) VALORACIÓN GLOBAL DEL MODELO
# ============================

porcentaje_R2 = R2 * 100
print("\n3a) Porcentaje de variabilidad explicada (R^2):", porcentaje_R2, "%")

modelo_valido = F_pval < 0.05
print("3b) ¿El modelo se considera válido globalmente?",
      "SÍ" if modelo_valido else "NO",
      "(p-valor F =", F_pval, ")")

```

```python
# EJERCICIO 4
prima_mercado = 0.06      # prima de riesgo del mercado
vol_mercado   = 0.20      # volatilidad del mercado

alpha1, beta1 = 0.0150, 1.050   # Empresa 1
alpha2, beta2 = 0.0060, 0.820   # Empresa 2

# 1) Prima de riesgo estimada para la empresa 1 (según CAPM)
risk_premium_emp1 = beta1 * prima_mercado
print("1) Prima de riesgo empresa 1:", risk_premium_emp1)

# 2) Riesgo sistemático de la empresa 2 (beta * sigma_M)
systematic_risk_emp2 = beta2 * vol_mercado
print("2) Riesgo sistemático empresa 2:", systematic_risk_emp2)

# 3) Beta de la cartera equiponderada de las dos empresas
beta_port = 0.5 * (beta1 + beta2)
print("3) Beta cartera equiponderada:", beta_port)

# 4) Primera implicación del CAPM: alpha_i = 0.
#    Miramos qué H0 (alpha=0) NO se rechaza con los p-valores de la tabla.
p_alpha1, p_alpha2 = 0.270, 0.038

no_rechaza_emp1 = p_alpha1 > 0.05   # no rechazamos alpha1=0 al 5%
no_rechaza_emp2 = p_alpha2 > 0.05   # sí rechazamos alpha2=0 al 5%

print("4) No se rechaza H0: alpha1 = 0 (empresa 1).")
print("   Para la empresa 2, alpha2 != 0 al 5%.")

# 5) Empresa para la que se cumple la primera implicación del CAPM
if no_rechaza_emp1 and not no_rechaza_emp2:
    print("5) La primera implicación del CAPM se cumple para la EMPRESA 1")

# 6) Ratio de Sharpe empresa 1 (desviación típica diaria 0.012)
sigma1 = 0.012
sharpe1 = risk_premium_emp1 / sigma1
print("6) Sharpe empresa 1:", sharpe1)

# 7) Ratio de Treynor empresa 2
risk_premium_emp2 = beta2 * prima_mercado
treynor2 = risk_premium_emp2 / beta2
print("7) Treynor empresa 2:", treynor2)

# 8) Alfa de Jensen empresa 2 si su rendimiento medio diario fue 0.007
#    Suponemos R_f ≈ 0, luego:
r_real_2 = 0.007
alpha_jensen_2 = r_real_2 - beta2 * prima_mercado
print("8) Alfa de Jensen empresa 2:", alpha_jensen_2)

# 9) Alfa de Jensen de la cartera equiponderada
#    (promedio de los alfas de ambas empresas)
alpha_port = 0.5 * (alpha1 + alpha2)
print("9) Alfa de Jensen cartera equiponderada:", alpha_port)

# 10) Sharpe de la cartera equiponderada si su desviación típica es 0.010
sigma_port = 0.010
risk_premium_port = beta_port * prima_mercado
sharpe_port = risk_premium_port / sigma_port
print("10) Sharpe cartera equiponderada:", sharpe_port)
```
