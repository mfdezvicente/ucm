## PREGUNTA 1

```python
#a) En un mercado se negocian los bonos a dos años A y B, con nominal 1000 y cupones del 10% y 5% respectivamente.

# Cada año se realizarán los pagos de los cupones que son C = 0.1*F para el A y C=0.05*F para el B. En el segundo año se devuelve también el principal


# b) Calcule los precios de los correspondientes bonos cupon cero a 1 y 2 años implícitos si los precios de los bonos A y B son PA=990 y PB=895: 

p_A = 990
p_B = 895 

# p_A = 100b1 + 1100b2
# p_B = 50b1 + 1050b2

# Debo resolver este sistema de ecuaciones. 


A = np.array([[100, 1100],
              [50, 1050]])

b = np.array([p_A, p_B])

# Resolver el sistema A·x = b
sol = np.linalg.solve(A, b)
b1, b2 = sol
print(f"El bono de cupón cero correspondiente al año 1 es b1 = {b1:.4f}")
print(f"El bono de cupón cero correspondiente al año 2 es b2 = {b2:.4f}")


#c) ¿Los precios p1 y p2 obtenidos son correctos? ¿por qué?
#El bono de cupón cero correspondiente al año 1 es b1 = 1.1000
#El bono de cupón cero correspondiente al año 2 es b2 = 0.8000
#Sí, los precios son correctos porque al sustituir los valores obtenidos en las ecuaciones originales se cumplen ambas igualdades.


#d) Determine las proporciones a invertir en los bonos A y B (xA y xB respectivamente) para replicar los pagos del bono cupón cero a un año

# Para un bono cupón cero a un año tendríamos que: 
# 1000 = 100x1 + 1100x2
# 0 = 50x1 + 1050x2
A_replica = np.array([[100, 50],
                      [1100, 1050]])


b_replica = np.array([1000, 0])
sol_replica = np.linalg.solve(A_replica, b_replica)
xA, xB = sol_replica
print(f"Proporción a invertir en el bono A (xA): {xA:.4f}")
print(f"Proporción a invertir en el bono B (xB): {xB:.4f}")

# e) Indique la corriente de pagos asociados a la siguiente estrategia de inversión: Vender un bono A y comprar w=xB/xA unidades del bono B (utiliza 6 decimales en los cálculos)
w = abs(xB / xA)
venta_A = -1  # Vender un bono A
compra_B = w  # Comprar w unidades del bono B
# Flow de pagos en el año 0, 1 y 2 de vender un bono A
pago_ano0_A = -venta_A * p_A
pago_ano1_A = venta_A * 100
pago_ano2_A = venta_A * 1100
# Flow de pagos en el año 0, 1 y 2 de comprar w unidades del bono B
pago_ano0_B = -compra_B * p_B
pago_ano1_B = compra_B * 50
pago_ano2_B = compra_B * 1050
# Corriente de pagos total
pago_ano0 = pago_ano0_A + pago_ano0_B
pago_ano1 = pago_ano1_A + pago_ano1_B
pago_ano2 = pago_ano2_A + pago_ano2_B

print(f"Corriente de pagos del año 0 de vender un bono A: {pago_ano0_A:.6f}")
print(f"Corriente de pagos del año 0 de comprar {w:.6f} unidades del bono B: {pago_ano0_B:.6f}")
print(f"Año 0: {pago_ano0:.6f}")

print(f"Corriente de pagos del año 1 de vender un bono A: {pago_ano1_A:.6f}")
print(f"Corriente de pagos del año 1 de comprar {w:.6f} unidades del bono B: {pago_ano1_B:.6f}")
print(f"Año 1: {pago_ano1:.6f}")

print(f"Corriente de pagos del año 2 de vender un bono A: {pago_ano2_A:.6f}")
print(f"Corriente de pagos del año 2 de comprar {w:.6f} unidades del bono B: {pago_ano2_B:.6f}")
print(f"Año 2: {pago_ano2:.6f}")

# f) Indique para qué valores del tipo de interés a un año sería esta estrategia rentable (valor actual de las ganancias positivo)
# Para que la estrategia sea rentable, el valor actual debe ser positivo
# VA > 0
print("El valor actual de las ganancias es positivo para valores de r1 tales que VA > 0.")
r = 0.0
while True:
    VA = pago_ano0 + pago_ano1 / (1 + r) + pago_ano2 / ((1 + r) ** 2)
    if VA > 0:
        break
    r += 0.0001
print(f"El valor actual de las ganancias es positivo para r1 > {r:.4f}")

```

```python
# Se ha estimado el siguiente modelo polinomial para la función de descuento (máxima precisión en los cálculos, mínimo 5 decimales)
# p(t) = 1− 0.01∗t −0.00099∗t^2 + 0.00003∗t^3
def p(t):
    return 1 - 0.01 * t - 0.00099 * t**2 + 0.00003 * t**3

# Determina el valor de la función de descuento a 6 meses, 2, 5 y 10 años:
tiempos = [0.5, 2, 5, 10]
for t in tiempos:
    valor_descuento = p(t)
    print(f"p({t} años) = {valor_descuento:.5f}")
 
#B. Determina el valor de los tipos de interés a 6 meses, 2, 5 y 10 años:
for t in tiempos:
    valor_descuento = p(t)
    tipo_interes = (1 / valor_descuento)**(1 / t) - 1
    print(f"Tipo de interés a {t} años: {tipo_interes:.5f}")

# Dibuja la curva de tipos de interés para t entre 0 y 10 años (utiliza un paso de 0.1 años)
import matplotlib.pyplot as plt
tiempos_curve = np.arange(0.1, 10.1, 0.1)
tipos_interes_curve = [(1 / p(t))**(1 / t) - 1 for t in tiempos_curve]
plt.plot(tiempos_curve, tipos_interes_curve)
plt.xlabel('Tiempo (años)')
plt.ylabel('Tipo de interés')
plt.title('Curva de tipos de interés')
plt.grid()
plt.show()

# d) Calcula los siguientes tipos forward:
# Entendemos por tipo forward f(t1, t2) al tipo de interés pactado hoy para un préstamo que comienza en t1 y finaliza en t2.
tiempos_forward = [(1, 2), (2, 5), (5, 10)]
for t1, t2 in tiempos_forward:
    p_t1 = p(t1)
    p_t2 = p(t2)
    tipo_forward = (p_t1 / p_t2)**(1 / (t2 - t1)) - 1
    print(f"Tipo forward f({t1}, {t2}) = {tipo_forward:.5f}")
    
```

```python
# Supón que lees en el periódico que los tipos de interés al contado correspondientes a 1, 2, 3, 4 y 5 años son del 5%, 4.75%, 4.5%, 4% y 3.75% respectivamente (pon las respuestas en tantos por uno redondeando a 5 decimales):

# a)  Calcula los precios de los siguientes bonos al descuento:
#  Bono a un año con nominal 1000: 
nominal = 1000
r1 = 0.05
p_bono_1 = nominal / (1 + r1)
print(f"El precio del bono a un año es: {p_bono_1:.5f}")

# Bono a dos años con nominal 900: 
nominal_2 = 900
r2 = 0.0475
p_bono_2 = nominal_2 / (1 + r2)**2
print(f"El precio del bono a dos años es: {p_bono_2:.5f}")

#  Bono a tres años con nominal 1: ç
nominal_3 = 1
r3 = 0.045
p_bono_3 = nominal_3 / (1 + r3)**3
print(f"El precio del bono a tres años es: {p_bono_3:.5f}")

# Bono a cuatro años con nominal 1500:
nominal_4 = 1500
r4 = 0.04
p_bono_4 = nominal_4 / (1 + r4)**4
print(f"El precio del bono a cuatro años es: {p_bono_4:.5f}")

# Bono a cinco años con nominal 1000:
nominal_5 = 1000
r5 = 0.0375
p_bono_5 = nominal_5 / (1 + r5)**5
print(f"El precio del bono a cinco años es: {p_bono_5:.5f}")



# b) Calcula el precio de los siguientes bonos con cupón, todos ellos con nominal 1000:

# Bono que paga 50 anualmente durante dos años: 
C = 50
nominal = 1000
r2 = 0.0475
p_bono_cupón_2 = C / (1 + r1) + (C + nominal) / (1 + r2)**2
print(f"El precio del bono con cupón a dos años es: {p_bono_cupón_2:.5f}")


# Bono que paga un cupón bienal (paga cada 2 años) de 200 durante 4 años 
C_bienal = 200
r4 = 0.04
p_bono_cupón_bienal = C_bienal / (1 + r2)**2 + (C_bienal + nominal) / (1 + r4)**4
print(f"El precio del bono con cupón bienal a cuatro años es: {p_bono_cupón_bienal:.5f}")

#  Bono a tres años con un cupón de 100.
C_3 = 100
r3 = 0.045
p_bono_cupón_3 = C_3 / (1 + r1) + C_3 / (1 + r2)**2 + (C_3 + nominal) / (1 + r3)**3
print(f"El precio del bono con cupón a tres años es: {p_bono_cupón_3:.5f}")


# Bono con un cupón anual del 5% y una tasa corriente del 6%.
C_anual = 0.05 * nominal
r_corriente = 0.06
p_bono_cupón_anual = (C_anual / (1 + r1) + C_anual / (1 + r2)**2 + C_anual / (1 + r3)**3 + C_anual / (1 + r4)**4 + (C_anual + nominal) / (1 + r5)**5)
print(f"El precio del bono con cupón anual del 5% es: {p_bono_cupón_anual:.5f}")



#C) Calcula la TIR de los bonos del apartado b): 

# TIR BONO 1. CUPON A DOS AÑOS DE 50 NOMINAL 1000
from scipy.optimize import newton
def funcion_tir_bono_2(tir):
    flujo_ano1 = 50 / (1 + tir)
    flujo_ano2 = 1050 / (1 + tir)**2
    return flujo_ano1 + flujo_ano2 - p_bono_cupón_2
tir_bono_cupón_2 = newton(funcion_tir_bono_2, 0.05)
print(f"La TIR del bono con cupón a dos años es: {tir_bono_cupón_2:.5f}")

# TIR BONO 2. CUPON BIENAL DE 200 NOMINAL 1000
def funcion_tir_bono_bienal(tir):
    flujo_ano2 = 200 / (1 + tir)**2
    flujo_ano4 = 1200 / (1 + tir)**4
    return flujo_ano2 + flujo_ano4 - p_bono_cupón_bienal
tir_bono_cupón_bienal = newton(funcion_tir_bono_bienal, 0.05)
print(f"La TIR del bono con cupón bienal a cuatro años es: {tir_bono_cupón_bienal:.5f}")

# TIR BONO 3. CUPON ANUAL DE 100 NOMINAL 1000
def funcion_tir_bono_3(tir):
    flujo_ano1 = 100 / (1 + tir)
    flujo_ano2 = 100 / (1 + tir)**2
    flujo_ano3 = 1100 / (1 + tir)**3
    return flujo_ano1 + flujo_ano2 + flujo_ano3 - p_bono_cupón_3
tir_bono_cupón_3 = newton(funcion_tir_bono_3, 0.05)
print(f"La TIR del bono con cupón a tres años es: {tir_bono_cupón_3:.5f}")

# TIR BONO 4. CUPON ANUAL DEL 5% NOMINAL 1000. SUPONEMOS QUE ES UN BONO A 2 AÑOS CON CUPONUAL
def funcion_tir_bono_4(tir):
    flujo_ano1 = 50 / (1 + tir)
    flujo_ano2 = 1050 / (1 + tir)**2
    return flujo_ano1 + flujo_ano2 - p_bono_cupón_anual
tir_bono_cupón_anual = newton(funcion_tir_bono_4, 0.05)
print(f"La TIR del bono con cupón anual del 5% es: {tir_bono_cupón_anual:.5f}")


# d) Calcular la Duración de los sigiuentes bonos. Utilizar los precios calculados en el apartado b) y tipos de interés al contado correspondientes a 1, 2, 3, 4 y 5 años son del 5%, 4.75%, 4.5%, 4% y 3.75% respectivamente.
tasas = [0.05, 0.0475, 0.045, 0.04, 0.0375]

#  Bono que paga 50 anualmente durante dos años:
def duracion_bono_cupón_2(precio, cupon, nominal, tasas):
    flujo_ano1 = cupon / (1 + tasas[0])
    flujo_ano2 = (cupon + nominal) / (1 + tasas[1])**2
    duracion = (1 * flujo_ano1 + 2 * flujo_ano2) / precio
    return duracion

duracion_1 = duracion_bono_cupón_2(p_bono_cupón_2, 50, 1000, tasas)
print(f"La duración del bono con cupón a dos años es: {duracion_1:.5f}")

# Bono que paga un cupón bienal de 200 durante 4 años
def duracion_bono_cupón_bienal(precio, cupon, nominal, tasas):
    flujo_ano2 = cupon / (1 + tasas[1])**2
    flujo_ano4 = (cupon + nominal) / (1 + tasas[3])**4
    duracion = (2 * flujo_ano2 + 4 * flujo_ano4) / precio
    return duracion

duracion_2 = duracion_bono_cupón_bienal(p_bono_cupón_bienal, 200, 1000, tasas)
print(f"La duración del bono con cupón bienal a cuatro años es: {duracion_2:.5f}")

# Bono a tres años con un cupón de 100:
def duracion_bono_cupón_3(precio, cupon, nominal, tasas):
    flujo_ano1 = cupon / (1 + tasas[0])
    flujo_ano2 = cupon / (1 + tasas[1])**2
    flujo_ano3 = (cupon + nominal) / (1 + tasas[2])**3
    duracion = (1 * flujo_ano1 + 2 * flujo_ano2 + 3 * flujo_ano3) / precio
    return duracion

duracion_3 = duracion_bono_cupón_3(p_bono_cupón_3, 100, 1000, tasas)
print(f"La duración del bono con cupón a tres años es: {duracion_3:.5f}")

# Cartera equiponderada compuesta por los tres bonos anteriores
def duracion_cartera_equiponderada(precios, cupones, nominales, tasas):
    duracion_total = 0
    for i in range(3):
        if i == 0:
            duracion_bono = duracion_bono_cupón_2(precios[i], cupones[i], nominales[i], tasas)
        elif i == 1:
            duracion_bono = duracion_bono_cupón_bienal(precios[i], cupones[i], nominales[i], tasas)
        else:
            duracion_bono = duracion_bono_cupón_3(precios[i], cupones[i], nominales[i], tasas)
        duracion_total += duracion_bono
    duracion_cartera = duracion_total / 3
    return duracion_cartera

# Duración de la cartera equiponderada
precios = [p_bono_cupón_2, p_bono_cupón_bienal, p_bono_cupón_3]
cupones = [50, 200, 100]
nominales = [1000, 1000, 1000]  
duracion_cartera_eq = duracion_cartera_equiponderada(precios, cupones, nominales, tasas)
print(f"La duración de la cartera equiponderada es: {duracion_cartera_eq:.5f}")

# Cartera inmunizada compuesta por posiciones largas en los primeros bonos del 50% cada una y una posición corta en el tercero:
def duracion_cartera_inmunizada(precios, cupones, nominales, tasas):
    duracion_bono1 = duracion_bono_cupón_2(precios[0], cupones[0], nominales[0], tasas)
    duracion_bono2 = duracion_bono_cupón_bienal(precios[1], cupones[1], nominales[1], tasas)
    duracion_bono3 = duracion_bono_cupón_3(precios[2], cupones[2], nominales[2], tasas)
    duracion_cartera = 0.5 * duracion_bono1 + 0.5 * duracion_bono2 - duracion_bono3
    return duracion_cartera

duracion_cartera_inm = duracion_cartera_inmunizada(precios, cupones, nominales, tasas)
print(f"La duración de la cartera inmunizada es: {duracion_cartera_inm:.5f}")

# Cartera formada por los tres bonos  inmunizada para asegurar que su valor dentro de 2.5 años será igual a 3000
def duracion_cartera_inmunizada_2(precios, cupones, nominales, tasas):
    duracion_bono1 = duracion_bono_cupón_2(precios[0], cupones[0], nominales[0], tasas)
    duracion_bono2 = duracion_bono_cupón_bienal(precios[1], cupones[1], nominales[1], tasas)
    duracion_bono3 = duracion_bono_cupón_3(precios[2], cupones[2], nominales[2], tasas)
    duracion_cartera = (duracion_bono1 + duracion_bono2 + duracion_bono3) / 3
    return duracion_cartera

duracion_cartera_inm2 = duracion_cartera_inmunizada_2(precios, cupones, nominales, tasas)
print(f"La duración de la cartera inmunizada para asegurar un valor de 3000 en 2.5 años es: {duracion_cartera_inm2:.5f}")



# e) E. Calcula la duración modificada de los tres primeros bonos del apartado anterior (utiliza la TIR de cada bono para ponderar) y la variación esperada en su precio ante un cambio de un 1% en TIR
def duracion_modificada(duracion, tir):
    return duracion / (1 + tir)
duracion_mod_1 = duracion_modificada(duracion_1, tir_bono_cupón_2)
duracion_mod_2 = duracion_modificada(duracion_2, tir_bono_cupón_bienal)
duracion_mod_3 = duracion_modificada(duracion_3, tir_bono_cupón_3)
print(f"La duración modificada del bono con cupón a dos años es: {duracion_mod_1:.4f}")
print(f"La duración modificada del bono con cupón bienal a cuatro años es: {duracion_mod_2:.4f}")
print(f"La duración modificada del bono con cupón a tres años es: {duracion_mod_3:.4f}")

# Calcula la variación esperada en su precio ante un cambio de un 1% en TIR
cambio_tir = 0.01
variacion_precio_1 = -duracion_mod_1 * cambio_tir * p_bono_cupón_2
variacion_precio_2 = -duracion_mod_2 * cambio_tir * p_bono_cupón_bienal
variacion_precio_3 = -duracion_mod_3 * cambio_tir * p_bono_cupón_3
print(f"La variación esperada en el precio del bono con cupón a dos años ante un cambio de 1% en TIR es: {variacion_precio_1:.4f}")
print(f"La variación esperada en el precio del bono con cupón bienal a cuatro años ante un cambio de 1% en TIR es: {variacion_precio_2:.4f}")
print(f"La variación esperada en el precio del bono con cupón a tres años ante un cambio de 1% en TIR es: {variacion_precio_3:.4f}")    







```
