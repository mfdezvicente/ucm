```python
# pip install pypdf

from pypdf import PdfReader, PdfWriter

input_pdf = "libro_sstt.pdf"
output_pdf = "libro_sstt_pares_180.pdf"

reader = PdfReader(input_pdf)
writer = PdfWriter()

for i, page in enumerate(reader.pages, start=1):  # 1..n
    if i % 2 == 0:  # páginas pares
        page.rotate(180)
    writer.add_page(page)

with open(output_pdf, "wb") as f:
    writer.write(f)

print(f"Guardado: {output_pdf}")

```
