## ¿Para qué sirve Pandas?

Pandas es tan versátil que quizás es más fácil decir lo que no puede hacer en lugar de lo que sí puede.


Por ejemplo, supongamos que deseas explorar un conjunto de datos almacenado en un CSV en tu ordenador. Pandas extraerá los datos de ese CSV a un DataFrame — básicamente una tabla — y luego te permitirá hacer cosas como:

- Calcular estadísticas y responder preguntas sobre los datos, como:


    - ¿Cuál es el promedio, mediana, máximo o mínimo de cada columna?
    - ¿La columna A se correlaciona con la columna B?
    - ¿Cómo es la distribución de datos en la columna C?


- Limpiar los datos haciendo cosas como eliminar valores faltantes y filtrar filas o columnas según algún criterio.


- Visualizar los datos con ayuda de Matplotlib. Crear gráficos de barras, líneas, histogramas, burbujas y más.


- Almacenar los datos limpios y transformados de nuevo en un CSV, otro archivo o base de datos.


Antes de saltar al modelado o las visualizaciones complejas, necesitas tener un buen entendimiento de la naturaleza de tu conjunto de datos y Pandas es la mejor vía para hacerlo.

## ¿Cómo encaja Pandas en el conjunto de herramientas de ciencia de datos?

La biblioteca Pandas no solo es un componente central del conjunto de herramientas de ciencia de datos, sino que se utiliza en conjunto con otras bibliotecas de esa colección.

Pandas está construido sobre el paquete **NumPy**, lo que significa que mucha de la estructura de NumPy se usa o se replica en Pandas. Los datos en Pandas a menudo se utilizan para alimentar el análisis estadístico en **SciPy**, las funciones de trazado de **Matplotlib** y los algoritmos de aprendizaje automático en **Scikit-learn**.

Los Jupyter Notebooks ofrecen un buen entorno para usar Pandas en la exploración de datos y modelado, pero Pandas también puede usarse en editores de texto con la misma facilidad.

Los dos componentes principales de pandas son `Series` y `DataFrame`.

Una `Series` es esencialmente una columna, y un `DataFrame` es una tabla multidimensional compuesta por una colección de Series.

Los DataFrames y las Series son bastante similares en que muchas operaciones que puedes hacer con uno, puedes hacerlas con el otro, como rellenar valores nulos y calcular la media.
